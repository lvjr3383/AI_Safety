# Section 1: Calculus for Everyone

## What is Calculus?
Calculus is simply the math of **change**.

- **Not** about fixed numbers (like algebra).
- **Not** about fixed shapes (like geometry).
- It's all about how things **move, grow, shrink, or transform**.

## Where Do We See Change?
Change is everywhere in our daily lives:
- **Your speed** as you drive.
- **Stock prices** going up and down.
- **Plant growth** over time.
- **AI models** getting better with each update.

Calculus helps us answer key questions about this change:
- How fast is something changing **at this exact moment**?
- Is it getting faster or slower?
- How can we find the best outcome?

## The Two Main Ideas

### 1. Differentiation (Finding "Speed")
Differentiation helps us understand **how fast something is changing right now**.
- Think of it as finding the "instantaneous speed" or "steepness" of a curve.
- **Why it matters for AI:** It tells us how much a model's error changes if we tweak its settings slightly.

### 2. Integration (Adding Up Changes)
Integration helps us **add up many tiny changes** to find a total.
- Think of it as finding the total distance traveled from speed, or the area under a curve.
- (We'll focus less on this in this notebook, but it's the other big part of calculus).

## Why Calculus is Key for AI
Modern AI, especially machine learning, relies heavily on calculus.
- When an AI model learns, it's constantly trying to reduce its "error."
- Calculus gives the AI the tools to figure out **how to adjust its internal parameters** most effectively to make better predictions.
- It helps the AI understand the "direction" and "amount" of change needed to learn optimally.

In this workbook, we'll dive into the ideas of differentiation, gradients, and optimization – showing you how these power AI from the ground up.

## Seeing Change in Action

To really grasp 'change,' let's look at some simple examples of how things can move over time. These graphs show different ways something's 'position' might change.

````python
import numpy as np
import matplotlib.pyplot as plt

# Section 1: Visualizing Change

time = np.linspace(0, 10, 100)

# Three different ways things can change
constant_speed = 2 * time
accelerating = 0.5 * time**2
slowing = 50 - 5 * np.exp(-time / 2)

fig, axes = plt.subplots(1, 3, figsize=(14, 4))

axes[0].plot(time, constant_speed, linewidth=2)
axes[0].set_title("Constant Speed")
axes[0].set_xlabel("Time")
axes[0].set_ylabel("Position")
axes[0].grid(True)

axes[1].plot(time, accelerating, linewidth=2)
axes[1].set_title("Accelerating")
axes[1].set_xlabel("Time")
axes[1].grid(True)

axes[2].plot(time, slowing, linewidth=2)
axes[2].set_title("Slowing Down")
axes[2].set_xlabel("Time")
axes[2].grid(True)

plt.tight_layout()
plt.show()

print("Observe how each graph changes:")
print("- Straight line → constant rate of change")
print("- Curved upward → increasing rate of change")
print("- Flattening curve → decreasing rate of change")
print("\nCalculus helps us measure these changes precisely.")
````

### What These Graphs Show:

*   **Constant Speed:** The straight line means something is changing at a steady pace, like a car on cruise control.
*   **Accelerating:** The curve getting steeper means the rate of change is increasing, like a car speeding up.
*   **Slowing Down:** The curve flattening out means the rate of change is decreasing, like a car braking.

Calculus gives us precise tools to measure these different types of change at any exact moment!

# Section 2: Functions

## What Is a Function?

A **function** is a rule that takes an **input** and produces an **output**.

Think of it as a machine.

```
 Input
   │
   ▼
┌──────────┐
│ Function │
└──────────┘
   │
   ▼
 Output
```

Every time you give the function an input, it follows the same rule and produces exactly one output.

For example,

```
f(x) = 2x + 1
```

If the input is:

- x = 1 → output = 3
- x = 2 → output = 5
- x = 5 → output = 11

The rule never changes.

---

## Functions Are Everywhere

You use functions every day without realizing it.

A few examples:

- 🌡️ A thermometer converts temperature into a reading.
- 🏦 A bank calculates interest from your balance.
- 🚗 A car's position changes as time passes.
- 🤖 A neural network converts inputs into predictions.

In each case:

**Input → Rule → Output**

---

## Function Notation

Functions are commonly written as:

```
f(x)
```

which is read as:

> "f of x"

Here:

- **x** is the input.
- **f** is the name of the function.
- **f(x)** is the output.

You may also see:

```
y = f(x)
```

This simply means that **y** is the value produced by the function.

---

## Why Functions Matter

Calculus is all about understanding **how functions change**.

Before we can measure change, we first need something that changes.

That "something" is a function.

````python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-4, 4, 200)

functions = [
    (2*x + 1, "f(x) = 2x + 1"),
    (x**2, "f(x) = x²"),
    (np.sin(x), "f(x) = sin(x)")
]

fig, axes = plt.subplots(1, 3, figsize=(13, 4))

for ax, (y, title) in zip(axes, functions):
    ax.plot(x, y, linewidth=2)
    ax.scatter([0, 1, 2],
               np.interp([0, 1, 2], x, y),
               s=40)
    ax.set_title(title)
    ax.set_xlabel("Input (x)")
    ax.set_ylabel("Output")
    ax.grid(True)

plt.tight_layout()
plt.show()
````

## Inputs and Outputs

Every function has two important parts.

### Independent Variable

The **independent variable** is the input.

We usually call it **x**.

You choose its value.

---

### Dependent Variable

The **dependent variable** is the output.

We usually call it **y**.

Its value depends on the input.

For example,

```
y = x²
```

If:

```
x = 3
```

then

```
y = 9
```

Change the input, and the output changes too.

---

This simple relationship between inputs and outputs forms the foundation of everything we'll study in calculus.

````python
x_values = [-2, -1, 0, 1, 2]

print("Input → Output")
print("-" * 18)

for x in x_values:
    y = x**2
    print(f"{x:>3}  →  {y}")
````

# Section 3: Reading Graphs

## What Does a Graph Tell Us?

A graph is simply a picture of a function.

Every point on the graph represents one input and one output.

```
Input (x) ─────► Output (y)
```

When we draw all of those input-output pairs together, we get a graph.

For example, if

```
f(x) = x²
```

then:

| Input (x) | Output (y) |
|-----------|-----------:|
| -2 | 4 |
| -1 | 1 |
| 0 | 0 |
| 1 | 1 |
| 2 | 4 |

Each row becomes a point on the graph.

Graphs help us see patterns that are difficult to notice from numbers alone.

Instead of reading a table, we can immediately answer questions like:

- Is the function going up or down?
- Is it changing quickly or slowly?
- Does the shape stay the same or curve?

Learning to read graphs is one of the most important skills in calculus.

````python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-2.5, 2.5, 200)
y = x**2

plt.figure(figsize=(6,4))
plt.plot(x, y, linewidth=2)

points = [(-2,4), (-1,1), (0,0), (1,1), (2,4)]
for px, py in points:
    plt.scatter(px, py, s=50)
    plt.text(px + 0.08, py + 0.2, f"({px}, {py})", fontsize=9)

plt.title("The Graph of f(x) = x²")
plt.xlabel("Input (x)")
plt.ylabel("Output (y)")
plt.grid(True)

plt.show()
````

## Increasing and Decreasing

As we move from left to right, a graph can behave in different ways.

### Increasing

An **increasing** function goes upward.

As the input gets larger, the output also gets larger.

📈 Think of money growing in a savings account.

---

### Decreasing

A **decreasing** function goes downward.

As the input gets larger, the output gets smaller.

📉 Think of a phone battery slowly draining during the day.

---

Some functions do both.

They may decrease for a while, then begin increasing again.

Learning to recognize these patterns is an important step toward understanding calculus.

````python
x = np.linspace(-3, 3, 200)

fig, axes = plt.subplots(1, 3, figsize=(13,4))

functions = [
    (x, "Always Increasing"),
    (-x, "Always Decreasing"),
    (x**2, "Decreasing then Increasing")
]

for ax, (y, title) in zip(axes, functions):
    ax.plot(x, y, linewidth=2)
    ax.set_title(title)
    ax.set_xlabel("x")
    ax.grid(True)

plt.tight_layout()
plt.show()
````

# Section 5: Slopes

## What Is a Slope?

Imagine walking up two different hills.

One hill has a gentle incline, making it easy to walk.

The other is much steeper and requires much more effort.

Even though both hills go upward, they don't rise at the same rate.

The mathematical way to describe this **steepness** is called the **slope**.

A slope tells us **how much the output changes when the input changes**.

For a straight line, the slope is calculated as:

\[
\text{Slope}=\frac{\text{Change in Output}}{\text{Change in Input}}
\]

or more commonly,

\[
\text{Slope}=\frac{\Delta y}{\Delta x}
\]

where:

- **Δy (rise)** is the vertical change.
- **Δx (run)** is the horizontal change.

For example, suppose two points are:

```
(1, 2)
(3, 6)
```

The output increased by:

```
6 − 2 = 4
```

The input increased by:

```
3 − 1 = 2
```

Therefore,

```
Slope = 4 ÷ 2 = 2
```

This means **for every 1-unit increase in x, y increases by 2 units.**

Slope gives us our first mathematical way of describing **how quickly something changes**.

````python
import numpy as np
import matplotlib.pyplot as plt

# Line passing through (1,2) and (3,6)
x = np.linspace(0, 4, 100)
y = 2 * x

plt.figure(figsize=(6,5))
plt.plot(x, y, linewidth=2)

# Points
plt.scatter([1, 3], [2, 6], color="red", s=60)

# Rise
plt.plot([3, 3], [2, 6], "--", linewidth=2)

# Run
plt.plot([1, 3], [2, 2], "--", linewidth=2)

plt.text(3.05, 4, "Rise = 4")
plt.text(2, 1.7, "Run = 2")

plt.title("Slope = Rise ÷ Run")
plt.xlabel("Input (x)")
plt.ylabel("Output (y)")
plt.grid(True)

plt.show()

print("Slope =", (6 - 2) / (3 - 1))
````

## Positive, Negative, and Zero Slopes

Not all lines behave the same way.

### Positive Slope

A positive slope means the graph rises as you move from left to right.

```
↗
```

The output increases as the input increases.

Example:

- Savings account growing
- Distance traveled while driving

---

### Negative Slope

A negative slope means the graph falls as you move from left to right.

```
↘
```

The output decreases as the input increases.

Example:

- Phone battery draining
- Fuel remaining during a road trip

---

### Zero Slope

A zero slope means the graph is perfectly flat.

```
──────
```

The output stays the same even though the input changes.

Example:

- A parked car
- A constant room temperature

The **sign** of the slope tells us the direction of change, while the **size** of the slope tells us how quickly that change happens.

````python
x = np.linspace(-3, 3, 100)

plt.figure(figsize=(7,5))

plt.plot(x, x, label="Positive Slope")
plt.plot(x, -x, label="Negative Slope")
plt.plot(x, np.full_like(x, 2), label="Zero Slope")

plt.title("Comparing Different Slopes")
plt.xlabel("Input (x)")
plt.ylabel("Output (y)")
plt.grid(True)
plt.legend()

plt.show()
````

## Straight Lines Are Easy...

Everything we've seen so far involves **straight lines**.

For straight lines, the slope is always the same.

No matter where you measure it, you always get the same answer.

But what happens if the graph is curved?

Consider this function:

```
y = x²
```

Near one point, the curve looks fairly flat.

At another point, it becomes much steeper.

Now the question becomes:

> **What is the slope at one exact point on the curve?**

The rise-over-run formula no longer gives a single answer because the slope keeps changing.

This question led to one of the greatest ideas in mathematics:

**the derivative.**

That's what we'll discover in the next section.

````python
x = np.linspace(-2.5, 2.5, 300)
y = x**2

plt.figure(figsize=(6,5))
plt.plot(x, y, linewidth=2)

point_x = 1
point_y = point_x**2

plt.scatter(point_x, point_y, color="red", s=70)
plt.annotate(
    "What is the slope here?",
    (point_x, point_y),
    xytext=(1.5, 4),
    arrowprops=dict(arrowstyle="->")
)

plt.title("A Curve Has a Different Steepness at Different Points")
plt.xlabel("Input (x)")
plt.ylabel("Output (y)")
plt.grid(True)

plt.show()
````

# Section 6: Derivatives

## What Is a Derivative?

Imagine you're driving a car.

At the end of your trip, someone tells you your **average speed** was 50 miles per hour.

That's useful—but it doesn't answer another important question:

> **How fast were you traveling at exactly 2:30 PM?**

Maybe you were stopped at a traffic light.

Maybe you were driving 70 mph on the highway.

The average speed can't tell us what happened at one specific moment.

Calculus was developed to answer questions like this.

Instead of asking:

> "How fast did something change overall?"

it asks:

> **"How fast is it changing right now?"**

That idea is called the **derivative**.

A derivative measures the **instantaneous rate of change** of a function.

Another way to think about it is:

> **The derivative tells us how steep a curve is at one exact point.**

Unlike a straight line, a curve can be flat in one place and steep in another.

The derivative lets us measure that changing steepness.

````python
import numpy as np
import matplotlib.pyplot as plt

# Function
x = np.linspace(-2, 4, 300)
y = x**2

# Point of interest
x0 = 1
y0 = x0**2

# Derivative of x² is 2x
slope = 2 * x0

# Tangent line
xt = np.linspace(x0 - 1.2, x0 + 1.2, 50)
yt = slope * (xt - x0) + y0

plt.figure(figsize=(6,5))
plt.plot(x, y, label="y = x²", linewidth=2)
plt.plot(xt, yt, "--", label="Tangent Line")
plt.scatter(x0, y0, color="red", s=60)

plt.annotate(
    "Derivative\n(slope here)",
    (x0, y0),
    xytext=(2, 5),
    arrowprops=dict(arrowstyle="->")
)

plt.title("The Derivative at One Point")
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True)
plt.legend()

plt.show()
````

## Positive, Negative, and Zero Derivatives

Because a derivative measures **slope**, it can be positive, negative, or zero.

### Positive Derivative

The function is rising.

As the input increases, the output increases.

📈

---

### Negative Derivative

The function is falling.

As the input increases, the output decreases.

📉

---

### Zero Derivative

The graph is perfectly flat at that point.

➡️

This often happens where the graph changes direction—for example, at the top of a hill or the bottom of a valley.

We'll explore these special points later when we study optimization.

````python
x = np.linspace(-3, 3, 400)
y = x**3 - 3*x

plt.figure(figsize=(7,5))
plt.plot(x, y, linewidth=2)

points = [-1, 0, 1]

for p in points:
    yp = p**3 - 3*p
    slope = 3*p**2 - 3

    xt = np.linspace(p - 0.4, p + 0.4, 20)
    yt = slope * (xt - p) + yp

    plt.plot(xt, yt, "--")
    plt.scatter(p, yp, s=50)

plt.title("The Derivative Changes Along a Curve")
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True)

plt.show()
````

## One Curve, Many Derivatives

A straight line has the same slope everywhere.

A curve is different.

As you move along the curve, its steepness changes.

That means the derivative also changes.

At one point, the derivative may be:

- Positive
- Negative
- Zero

The derivative isn't a single number that belongs to the whole graph.

Instead, every point on the graph has its own derivative.

This is one of the most important ideas in calculus.

# Section 7: Estimating Derivatives with Python

## How Can We Calculate a Derivative?

In the previous section, we learned that a **derivative** measures the slope of a curve at one exact point.

But that raises an important question:

> **How can we actually calculate it?**

Imagine you're watching a runner.

- If you measure their position one hour apart, you only get an average speed.
- Measure one minute apart, and your estimate improves.
- Measure one second apart, and it's even better.

The closer together your measurements are, the closer you get to the runner's **instantaneous speed**.

We can use exactly the same idea for functions.

Instead of measuring one point, we measure **two points that are very close together**.

This approach is called **finite differences**.

For a function \(f(x)\), we estimate the derivative using:

\[
\frac{f(x+h)-f(x)}{h}
\]

where:

- **x** is the point we're interested in.
- **h** is a very small number.

The smaller **h** becomes, the closer our estimate gets to the true derivative.

Let's see this idea in Python.

````python
import numpy as np
import matplotlib.pyplot as plt

def numerical_derivative(f, x, h=1e-5):
    return (f(x + h) - f(x)) / h

# Function: f(x) = x²
f = lambda x: x**2

x = 1

estimate = numerical_derivative(f, x)

print(f"Estimated derivative at x = {x}: {estimate:.6f}")
print(f"True derivative: {2*x}")
````

## Better Estimates with Smaller Steps

The value of **h** controls how far apart our two points are.

Suppose we estimate the derivative of

\[
f(x)=x^2
\]

at **x = 1**.

If:

- **h = 0.1**, the estimate is fairly good.
- **h = 0.01**, it's even closer.
- **h = 0.001**, it's extremely close.

As **h** gets smaller, our approximation approaches the true slope of the curve.

This connects directly to what we saw in the previous section.

Instead of drawing a tangent line by eye, we're letting the computer estimate its slope using two nearby points.

In practice, computers choose values of **h** that are small enough to produce accurate results without running into rounding errors.

````python
h_values = [0.1, 0.01, 0.001, 0.0001]

true_value = 2

print(f"{'h':<10}{'Estimated Derivative':<25}{'Error'}")
print("-" * 50)

for h in h_values:
    estimate = numerical_derivative(f, 1, h)
    error = abs(estimate - true_value)
    print(f"{h:<10}{estimate:<25.6f}{error:.8f}")
````

````python
h_values = np.array([0.1, 0.01, 0.001, 0.0001])
estimates = [numerical_derivative(f, 1, h) for h in h_values]

plt.figure(figsize=(6,4))
plt.plot(h_values, estimates, "o-", label="Estimated Derivative")
plt.axhline(2, color="red", linestyle="--", label="True Derivative")

plt.xscale("log")
plt.gca().invert_xaxis()

plt.title("Smaller h Gives a Better Estimate")
plt.xlabel("h (log scale)")
plt.ylabel("Derivative Estimate")
plt.grid(True)
plt.legend()

plt.show()
````

# Section 8: Common Derivative Patterns

## Mathematics Loves Patterns

One of the beautiful things about mathematics is that it is full of patterns.

After studying thousands of functions, mathematicians discovered that derivatives are **not random**.

They follow consistent rules.

That means once you recognize a pattern, you can often predict the derivative without starting from scratch.

The goal of this section is **not** to memorize formulas.

Instead, it's to build intuition for the patterns you'll see again and again.

We'll start with the most important rule in introductory calculus.

## The Power Rule

Consider these functions:

| Function | Derivative |
|----------|-----------|
| \(x\) | \(1\) |
| \(x^2\) | \(2x\) |
| \(x^3\) | \(3x^2\) |
| \(x^4\) | \(4x^3\) |

Notice the pattern.

1. The exponent comes down in front.
2. The exponent decreases by one.

This gives us the **Power Rule**:

\[
\frac{d}{dx}(x^n)=n x^{\,n-1}
\]

This single rule allows us to differentiate a huge number of functions.

Don't worry about memorizing it immediately.

After using it a few times, the pattern becomes surprisingly natural.

````python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-2, 2, 300)

plt.figure(figsize=(7,5))

plt.plot(x, x, label=r"$x$")
plt.plot(x, x**2, label=r"$x^2$")
plt.plot(x, x**3, label=r"$x^3$")
plt.plot(x, x**4, label=r"$x^4$")

plt.title("Functions That Follow the Power Rule")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.grid(True)
plt.legend()

plt.show()

print("Power Rule:")
print("x    → 1")
print("x²   → 2x")
print("x³   → 3x²")
print("x⁴   → 4x³")
````

## A Very Special Function: \(e^x\)

Most functions change when we take their derivative.

For example:

- \(x^2\) becomes \(2x\)
- \(x^3\) becomes \(3x^2\)

But one function is different.

\[
e^x
\]

Its derivative is simply:

\[
\frac{d}{dx}(e^x)=e^x
\]

In other words, **it is its own derivative**.

This remarkable property makes the exponential function incredibly useful in science, engineering, probability, and machine learning.

````python
x = np.linspace(-2, 2, 300)

y = np.exp(x)

plt.figure(figsize=(6,4))

plt.plot(x, y, linewidth=2, label=r"$e^x$")
plt.plot(x, y, "--", linewidth=2, label="Derivative")

plt.title(r"$e^x$ and Its Derivative")
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True)
plt.legend()

plt.show()
````

## The Natural Logarithm

The natural logarithm, written as

\[
\ln(x)
\]

is closely related to the exponential function.

In fact, it "undoes" the exponential function, making the two mathematical inverses of each other.

Its derivative follows another elegant pattern:

\[
\frac{d}{dx}\ln(x)=\frac{1}{x}
\]

You'll encounter logarithms frequently in probability, statistics, information theory, and machine learning.

Like the Power Rule, this derivative appears so often that it eventually becomes familiar through practice.

````python
summary = [
    ("x", "1"),
    ("x²", "2x"),
    ("x³", "3x²"),
    ("x⁴", "4x³"),
    ("eˣ", "eˣ"),
    ("ln(x)", "1/x"),
]

print(f"{'Function':<12}Derivative")
print("-" * 24)

for function, derivative in summary:
    print(f"{function:<12}{derivative}")
````

## What If a Function Has More Than One Input?

So far, most of our functions looked like this:

\[

\]

One input goes in.

One output comes out.

But in the real world, many things depend on **more than one input**.

Examples:

| Situation | Inputs | Output |

````python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-3, 3, 80)
y = np.linspace(-3, 3, 80)

X, Y = np.meshgrid(x, y)
Z = X**2 + Y**2

fig = plt.figure(figsize=(7, 5))
ax = fig.add_subplot(111, projection="3d")

ax.plot_surface(X, Y, Z, alpha=0.8)

ax.set_title("Surface: f(x, y) = x² + y²")
ax.set_xlabel("x")
ax.set_ylabel("y")
ax.set_zlabel("f(x, y)")

plt.show()
````

## Reading a 3D Surface

The graph above shows every possible pair of inputs:

\[
(x, y)

and the output:

f(x, y)

f(x, y)=x^2+y^2

\[

# Section 10: Partial Derivatives

## Changing One Input at a Time

In the previous section, we looked at functions with more than one input.

For example,

\[
f(x,y)=x^2+y^2
\]

Now imagine you're standing on the surface of this function.

You can move in different directions.

- Move in the **x-direction** while keeping **y** fixed.
- Move in the **y-direction** while keeping **x** fixed.

Each direction has its own slope.

Instead of asking,

> "How does the function change?"

we ask a more specific question:

> **"How does the function change if only one input changes?"**

The answer is called a **partial derivative**.

A partial derivative measures how the output changes when **one input changes while all the others stay the same**.

This idea is one of the foundations of multivariable calculus.

## Partial Derivative Notation

Partial derivatives have their own notation.

Instead of writing

\[
\frac{d}{dx}
\]

we write

\[
\frac{\partial}{\partial x}
\]

The curly symbol

\[
\partial
\]

(pronounced "partial")

reminds us that we're changing **only one variable**.

For the function

\[
f(x,y)=x^2+y^2
\]

the partial derivatives are:

With respect to **x**:

\[
\frac{\partial f}{\partial x}=2x
\]

because we treat **y** as a constant.

With respect to **y**:

\[
\frac{\partial f}{\partial y}=2y
\]

because we treat **x** as a constant.

The same function has two different partial derivatives because we are measuring change in two different directions.

````python
import numpy as np

def partial_x(f, x, y, h=1e-5):
    return (f(x + h, y) - f(x, y)) / h

def partial_y(f, x, y, h=1e-5):
    return (f(x, y + h) - f(x, y)) / h

f = lambda x, y: x**2 + y**2

x0, y0 = 2, 1

print(f"Point: ({x0}, {y0})")
print(f"∂f/∂x ≈ {partial_x(f, x0, y0):.5f}")
print(f"∂f/∂y ≈ {partial_y(f, x0, y0):.5f}")
````

## A Simple Pattern

The good news is that the derivative rules you've already learned still work.

The only difference is this:

> **Treat every other variable as a constant.**

For example,

\[
f(x,y)=3x^2y
\]

To find the partial derivative with respect to **x**, treat **y** like an ordinary number.

\[
\frac{\partial f}{\partial x}=6xy
\]

Now do the opposite.

Treat **x** like a constant.

\[
\frac{\partial f}{\partial y}=3x^2
\]

The mathematics is familiar—you simply focus on one variable at a time.

````python
import matplotlib.pyplot as plt

x = np.linspace(-3, 3, 80)
y = np.linspace(-3, 3, 80)

X, Y = np.meshgrid(x, y)
Z = X**2 + Y**2

fig = plt.figure(figsize=(7,5))
ax = fig.add_subplot(111, projection="3d")

ax.plot_surface(X, Y, Z, alpha=0.6)

x0, y0 = 1, 2
z0 = x0**2 + y0**2

ax.scatter(x0, y0, z0, color="red", s=60)

ax.text(x0, y0, z0 + 2,
        "Different slopes\nin different directions",
        color="red")

ax.set_title("Partial Derivatives at One Point")
ax.set_xlabel("x")
ax.set_ylabel("y")
ax.set_zlabel("f(x,y)")

plt.show()
````

# Section 11: Gradients

## What Is a Gradient?

In the previous section, we learned about **partial derivatives**.

For a function with two inputs:

\[
f(x, y) = x^2 + y^2
\]

we can ask:

- How does the output change if **x** changes? (∂f/∂x)
- How does the output change if **y** changes? (∂f/∂y)

But a more powerful question is:

> **"In what direction should I move to improve the function the fastest?"**

This question is answered by combining all the partial derivatives into a single mathematical object called the **gradient**.

The **gradient** is a vector that points in the direction of steepest increase.

It collects all partial derivatives together:

\[
\nabla f = \left[ \frac{\partial f}{\partial x}, \frac{\partial f}{\partial y} \right]
\]

The symbol ∇ (pronounced "nabla" or "del") represents the gradient.

For our example:

\[
\nabla f = [2x, 2y]
\]

This vector has two pieces of information:

- **Direction:** Which way should we move?
- **Magnitude:** How steep is the slope in that direction?

````python
import numpy as np
import matplotlib.pyplot as plt

# Define the function
f = lambda x, y: x**2 + y**2

# Partial derivatives
partial_x = lambda x, y: 2*x
partial_y = lambda x, y: 2*y

# Create a grid of points
x = np.linspace(-3, 3, 10)
y = np.linspace(-3, 3, 10)
X, Y = np.meshgrid(x, y)

# Compute gradients at each point
GX = partial_x(X, Y)
GY = partial_y(X, Y)

# Plot the contours
fig, ax = plt.subplots(figsize=(7, 6))

contours = ax.contour(X, Y, f(X, Y), levels=15, alpha=0.6)

# Plot gradient vectors
ax.quiver(X, Y, GX, GY, alpha=0.7, scale=30)

ax.set_xlabel("x")
ax.set_ylabel("y")
ax.set_title("Gradient Vectors (Arrows Point Uphill)")
ax.grid(True, alpha=0.3)

plt.show()

# Example: compute gradient at one point
x0, y0 = 1.5, 1.0
grad_x = partial_x(x0, y0)
grad_y = partial_y(x0, y0)

print(f"At point ({x0}, {y0}):")
print(f"∇f = [{grad_x}, {grad_y}]")
print(f"Magnitude: {np.sqrt(grad_x**2 + grad_y**2):.3f}")
````

## Gradient Points Uphill

The gradient always points in the direction of **steepest increase**.

If you follow the gradient, you climb the surface as steeply as possible.

But for optimization, we want the opposite.

We want to **minimize** the function (reach the lowest point).

So we follow the **negative gradient**:

\[
-\nabla f
\]

The negative gradient points **downhill**.

This is the key insight that makes optimization work.

If we repeatedly take small steps in the direction of the negative gradient, we slowly descend toward lower and lower values.

This simple idea is the foundation of nearly all machine learning training algorithms.

````python
# Visualize gradient and negative gradient at specific points

fig, ax = plt.subplots(figsize=(8, 7))

# Contour plot
x_fine = np.linspace(-3, 3, 100)
y_fine = np.linspace(-3, 3, 100)
X_fine, Y_fine = np.meshgrid(x_fine, y_fine)
Z_fine = f(X_fine, Y_fine)

contours = ax.contour(X_fine, Y_fine, Z_fine, levels=15, alpha=0.5)

# A few selected points to show gradient
points = [(2, 2), (1.5, -1.5), (-1, 1.5)]
colors = ['red', 'blue', 'green']

for (px, py), color in zip(points, colors):
    grad_x = partial_x(px, py)
    grad_y = partial_y(px, py)

    # Normalize for visibility
    magnitude = np.sqrt(grad_x**2 + grad_y**2)
    grad_x_norm = grad_x / magnitude * 0.5
    grad_y_norm = grad_y / magnitude * 0.5

    # Gradient (uphill)
    ax.arrow(px, py, grad_x_norm, grad_y_norm,
             head_width=0.15, head_length=0.1, fc=color, ec=color, label=f'∇f at ({px},{py})')

    # Negative gradient (downhill)
    ax.arrow(px, py, -grad_x_norm*0.5, -grad_y_norm*0.5,
             head_width=0.1, head_length=0.08, fc=color, ec=color,
             linestyle='--', alpha=0.5)

ax.set_xlabel("x")
ax.set_ylabel("y")
ax.set_title("Gradient (solid) Points Uphill | Negative Gradient (dashed) Points Downhill")
ax.set_xlim(-3, 3)
ax.set_ylim(-3, 3)
ax.grid(True, alpha=0.3)
ax.legend()

plt.show()
````

## Gradient Magnitude Matters

The magnitude of the gradient tells us **how steep** the surface is.

Large magnitude = steep slope.

Small magnitude = gentle slope.

At the center of our bowl-shaped function (0, 0), both partial derivatives are zero:

\[
\nabla f = [0, 0]
\]

This is called a **critical point**.

The magnitude is zero because the surface is perfectly flat there—there's no direction to improve.

Away from the center, the gradient gets larger, indicating steeper terrain.

This information is crucial for optimization:

- Where the gradient is large, we can improve quickly.
- Where the gradient is small, we're near a critical point.

# Section 12: The Chain Rule

## Functions Inside Functions

Imagine a simple process:

1. You square a number.
2. You add 3 to the result.

In mathematical notation:

x → square it → x² → add 3 → x² + 3

This is called a **composite function**—one function feeds into another.

More formally:
f(g(x))

where:

- **g(x)** is the inner function (what happens first)
- **f** is the outer function (what happens to the result)

For example:
g(x) = x²
f(u) = u + 3

So:
f(g(x)) = x² + 3

The key question is:

> **If x changes, how much does the final output change?**

The answer requires understanding how change flows through the chain.

This is what the **chain rule** answers.

````python
import numpy as np
import matplotlib.pyplot as plt

# Composite function: first square, then add 3
x = np.linspace(-2, 2, 100)
inner = x**2
outer = inner + 3

# Let's trace through one example
x_example = 1.5
inner_example = x_example**2
outer_example = inner_example + 3

print("Example: x =", x_example)
print(f"Step 1 (inner): x² = {inner_example}")
print(f"Step 2 (outer): x² + 3 = {outer_example}")

# Plot the composite function
plt.figure(figsize=(6, 4))
plt.plot(x, outer, linewidth=2)
plt.scatter(x_example, outer_example, color='red', s=60)
plt.annotate(
    f"({x_example}, {outer_example})",
    (x_example, outer_example),
    xytext=(0.5, 5),
    arrowprops=dict(arrowstyle="->")
)

plt.title("Composite Function: f(g(x)) = x² + 3")
plt.xlabel("x")
plt.ylabel("f(g(x))")
plt.grid(True)
plt.show()
````

## How Change Flows Through the Chain

Imagine you nudge **x** slightly.

That nudge affects the inner function.

The inner change then affects the outer function.

The total change at the end is the **product** of these individual changes.

Here's the key insight:

Total change = (change in outer) × (change in inner)

Or in calculus notation:
d/dx[f(g(x))] = f'(g(x)) · g'(x)

This is saying:

- **f'(g(x))** is how much the outer function changes when its input changes.
- **g'(x)** is how much the inner function changes when x changes.
- Multiply them together to get the total effect.

Think of it like a chain reaction.

A small change at the beginning gets amplified (or dampened) by each link in the chain.

## The Chain Rule in Action

Let's use a concrete example:

\[
y = (x^2 + 1)^3
\]

**Step 1: Identify the inner and outer functions**

- Inner: \(g(x) = x^2 + 1\)
- Outer: \(f(u) = u^3\)

**Step 2: Find the derivatives**

- Derivative of inner: \(g'(x) = 2x\)
- Derivative of outer: \(f'(u) = 3u^2\)

**Step 3: Apply the chain rule**

\[
\frac{dy}{dx} = f'(g(x)) \cdot g'(x) = 3(x^2 + 1)^2 \cdot 2x
\]

Simplify:

\[
\frac{dy}{dx} = 6x(x^2 + 1)^2
\]

That's it.

The chain rule lets us find the derivative of composite functions systematically.

We don't need to expand the entire expression—we just multiply the derivatives at each stage.

````python
def numerical_derivative(f, x, h=1e-5):
    return (f(x + h) - f(x)) / h

# Function: y = (x² + 1)³
f = lambda x: (x**2 + 1)**3

# Chain rule prediction:
# dy/dx = 6x(x² + 1)²
chain_rule = lambda x: 6 * x * (x**2 + 1)**2

# Test at a few points
test_points = [0, 1, 2]

print(f"{'x':<5}{'Chain Rule':<20}{'Numerical':<20}{'Error'}")
print("-" * 60)

for x in test_points:
    predicted = chain_rule(x)
    estimated = numerical_derivative(f, x)
    error = abs(predicted - estimated)
    print(f"{x:<5}{predicted:<20.6f}{estimated:<20.6f}{error:.8f}")

print("\nThe chain rule matches the numerical derivative!")
````

## Why This Matters for Deep Learning

A neural network is a **very long chain** of functions.

Input → Layer 1 → Layer 2 → Layer 3 → ... → Output → Loss

Each layer is a function.

The output of one layer becomes the input of the next.

If we change a weight in Layer 1, that change propagates through Layer 2, then Layer 3, and finally affects the loss.

The question is:

> **How much did that weight in Layer 1 contribute to the final loss?**

The answer requires applying the chain rule many times.

We go backward from the loss, through each layer, computing the effect at each step.

This process is called **backpropagation**, and it's the foundation of how neural networks learn.

Without the chain rule, we couldn't train deep networks at all.

# Section 13: Minima, Maxima & Curvature

## Climbing Mountains and Descending Into Valleys

Imagine you're hiking through mountains.

Sometimes you reach the **top of a hill**—a peak.

Sometimes you find yourself in a **valley**—the lowest point around.

Sometimes you're on a **flat plateau** where the ground doesn't change.

In the language of calculus:

- **Local maximum:** A peak. The function is higher here than at nearby points.
- **Local minimum:** A valley. The function is lower here than at nearby points.
- **Saddle point or inflection:** A flat region that isn't quite a peak or valley.

All of these special points share something in common:

At each one, the derivative is **zero**.

The tangent line is perfectly flat.

But here's the catch:

A zero derivative doesn't always mean you've found a peak or valley. Sometimes it means something else entirely.

In this section, we'll learn to recognize the difference.

````python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-2.5, 2.5, 300)
y = x**3 - 3*x

plt.figure(figsize=(7, 5))
plt.plot(x, y, linewidth=2)

# Mark critical points (where derivative = 0)
# Derivative of x³ - 3x is 3x² - 3 = 0 at x = ±1
plt.scatter([-1, 1], [2, -2], color='red', s=100, zorder=5)

plt.annotate('Local Maximum', xy=(-1, 2), xytext=(-1.8, 3.5),
             arrowprops=dict(arrowstyle='->', color='red'))
plt.annotate('Local Minimum', xy=(1, -2), xytext=(1.5, -3.5),
             arrowprops=dict(arrowstyle='->', color='red'))

plt.title('Finding Peaks and Valleys')
plt.xlabel('x')
plt.ylabel('y')
plt.grid(True, alpha=0.3)
plt.axhline(0, color='k', linewidth=0.5)
plt.axvline(0, color='k', linewidth=0.5)

plt.show()

print("Critical points (where derivative = 0):")
print("x = -1: Local maximum at y = 2")
print("x = 1: Local minimum at y = -2")
````

## What Does the Derivative Tell Us?

The derivative tells us whether the function is rising or falling.

**Positive derivative:** The function is going up. As x increases, y increases.

**Negative derivative:** The function is going down. As x increases, y decreases.

**Zero derivative:** The function is flat. It's either at a peak, a valley, or pausing briefly.

Here's the key insight:

When the derivative **changes sign**, something important happens.

- If it goes from **positive to zero to negative**, you've found a **maximum** (a peak).
- If it goes from **negative to zero to positive**, you've found a **minimum** (a valley).
- If it stays the same sign on both sides, the zero derivative doesn't represent a peak or valley—it's something else.

This is how we recognize special points: not by looking at the function itself, but by watching how its derivative changes.

````python
x = np.linspace(-2.5, 2.5, 300)
y = x**3 - 3*x

# Derivative: 3x² - 3
y_prime = 3*x**2 - 3

fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(7, 8))

# Original function
ax1.plot(x, y, linewidth=2, color='blue')
ax1.scatter([-1, 1], [2, -2], color='red', s=100, zorder=5)
ax1.axhline(0, color='k', linewidth=0.5)
ax1.axvline(0, color='k', linewidth=0.5)
ax1.set_title('Original Function: y = x³ - 3x')
ax1.set_ylabel('y')
ax1.grid(True, alpha=0.3)

# Derivative
ax2.plot(x, y_prime, linewidth=2, color='green')
ax2.axhline(0, color='red', linewidth=2, linestyle='--', label='Derivative = 0')
ax2.scatter([-1, 1], [0, 0], color='red', s=100, zorder=5)
ax2.set_title("Derivative: y' = 3x² - 3")
ax2.set_xlabel('x')
ax2.set_ylabel("y'")
ax2.grid(True, alpha=0.3)
ax2.legend()

# Annotate regions
ax2.text(-2, 8, 'Positive\n(rising)', fontsize=10, color='green')
ax2.text(0, 8, 'Negative\n(falling)', fontsize=10, color='green')
ax2.text(2, 8, 'Positive\n(rising)', fontsize=10, color='green')

plt.tight_layout()
plt.show()
````

Imagine standing on a road.

The road can have different shapes:

**🙂 A bowl (valley):** If you stand at the bottom and look up, both sides go upward.

**🙃 An upside-down bowl (peak):** If you stand at the top and look down, both sides go downward.

**→ A straight road:** The ground doesn't curve at all.

This shape is called **curvature**.

Curvature tells us something important about critical points:

**Positive curvature:** The shape is like a bowl. The zero derivative points to a **minimum**.

You don't need formulas to understand this.

Just look at the shape:

- If the graph curves upward, it's a minimum.
- If the graph curves downward, it's a maximum.

This simple visual intuition is enough to recognize peaks and valleys.

````python
x = np.linspace(-3, 3, 300)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 4))

# Minimum: y = x²
y1 = x**2
ax1.plot(x, y1, linewidth=2, color='blue')
ax1.scatter([0], [0], color='red', s=100, zorder=5)
ax1.set_title('Positive Curvature\n(Bowl - Minimum)', fontsize=12)
ax1.set_xlabel('x')
ax1.set_ylabel('y')
ax1.grid(True, alpha=0.3)
ax1.set_ylim(-1, 10)
ax1.text(-2, 7, '🙂', fontsize=40)

# Maximum: y = -x²
y2 = -x**2
ax2.plot(x, y2, linewidth=2, color='red')
ax2.scatter([0], [0], color='red', s=100, zorder=5)
ax2.set_title('Negative Curvature\n(Hill - Maximum)', fontsize=12)
ax2.set_xlabel('x')
ax2.set_ylabel('y')
ax2.grid(True, alpha=0.3)
ax2.set_ylim(-10, 1)
ax2.text(-2, -7, '🙃', fontsize=40)

plt.tight_layout()
plt.show()

print("y = x²:  curves upward → minimum at x=0")
print("y = -x²: curves downward → maximum at x=0")
````

# Section 14: Loss Functions

## What Are We Trying to Minimize?

Throughout this workbook, we've talked about finding the lowest point of a function.

But a crucial question remains:

> **Lowest point of what?**

The answer is the **loss function** (also called the **cost function** or **error function**).

A loss function measures how wrong a model's predictions are.

It takes two inputs:

- **Predicted values:** What the model thinks the answer is.
- **Actual values:** What the answer really is.

It outputs a single number:

- **The loss:** A measure of how far off the predictions are.

The goal of training is simple:

Make the loss as small as possible.

A small loss means accurate predictions.

A large loss means inaccurate predictions.

That's it. Everything in machine learning boils down to minimizing this one number.

````python
import numpy as np
import matplotlib.pyplot as plt

# Example: predicting house prices
actual_prices = np.array([300, 250, 400, 350])
predicted_prices = np.array([310, 240, 420, 330])

# Calculate errors (differences)
errors = predicted_prices - actual_prices

print("Actual prices:", actual_prices)
print("Predicted prices:", predicted_prices)
print("Errors:", errors)
print()

# Visualize
fig, ax = plt.subplots(figsize=(8, 4))

x_pos = np.arange(len(actual_prices))
width = 0.35

ax.bar(x_pos - width/2, actual_prices, width, label='Actual', alpha=0.8)
ax.bar(x_pos + width/2, predicted_prices, width, label='Predicted', alpha=0.8)

ax.set_xlabel('House')
ax.set_ylabel('Price ($1000s)')
ax.set_title('Predictions vs Actual Values')
ax.set_xticks(x_pos)
ax.set_xticklabels(['House 1', 'House 2', 'House 3', 'House 4'])
ax.legend()
ax.grid(True, alpha=0.3, axis='y')

plt.tight_layout()
plt.show()

# The errors are: [10, -10, 20, -20]
# But how do we combine them into ONE number (the loss)?
````

## Common Loss Functions

There are many ways to measure error. Here are the most important ones.

### Mean Squared Error (MSE)

Square each error, then take the average.

Formula:

\[
\text{MSE} = \frac{1}{n} \sum_{i=1}^{n} (y_{\text{predicted}} - y_{\text{actual}})^2
\]

**Why squaring?**

- Squaring makes negative errors positive (so -10 and +10 both contribute equally).
- Large errors are penalized more heavily (squaring amplifies big mistakes).
- MSE is smooth and works well with calculus (easy to differentiate).

### Mean Absolute Error (MAE)

Take the absolute value of each error, then average.

Formula:

\[
\text{MAE} = \frac{1}{n} \sum_{i=1}^{n} |y_{\text{predicted}} - y_{\text{actual}}|
\]

**Why absolute value?**

- Simpler interpretation: it's the average error magnitude.
- Treats all errors equally (a $10 mistake is always $10, not amplified).
- More robust to outliers than MSE.

### Which One to Use?

- **MSE:** When you want to penalize large errors heavily. Common for regression.
- **MAE:** When you want a simpler, more interpretable measure. Also good for regression.
- Both are used in practice. The choice depends on your specific problem.

````python
# Calculate MSE and MAE for our example

actual = np.array([300, 250, 400, 350])
predicted = np.array([310, 240, 420, 330])

errors = predicted - actual

# Mean Squared Error
mse = np.mean(errors**2)

# Mean Absolute Error
mae = np.mean(np.abs(errors))

print(f"Mean Squared Error (MSE): {mse}")
print(f"Mean Absolute Error (MAE): {mae}")
print()

# Visualize how they penalize errors differently
error_range = np.linspace(-30, 30, 100)

squared_errors = error_range**2
absolute_errors = np.abs(error_range)

fig, ax = plt.subplots(figsize=(7, 5))

ax.plot(error_range, squared_errors, linewidth=2, label='Squared Error (MSE)')
ax.plot(error_range, absolute_errors, linewidth=2, label='Absolute Error (MAE)')

ax.set_xlabel('Prediction Error')
ax.set_ylabel('Penalty')
ax.set_title('MSE vs MAE: How They Penalize Errors')
ax.legend()
ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

print("Notice: MSE curves upward sharply (large errors are heavily penalized)")
print("Notice: MAE is linear (all errors penalized equally)")
````

The loss function defines what "good" means for your model.

**Example 1: Predicting House Prices**

If your predictions are off by $10,000 on a $300,000 house, that's a problem.

**MSE is good here** because it penalizes big mistakes.

**Example 2: Counting Objects in Images**

If you predict 15 objects but there are 14, that's a small error.

If you predict 5 objects but there are 20, that's a big error.

You might use **MAE** because the error magnitude is what matters.

**Example 3: Classification (Spam vs Not Spam)**

Different loss functions exist for classification problems (like **cross-entropy**).

These we'll explore later.

The key idea: The loss function encodes what matters for your problem.

During training, the model learns to minimize whatever loss function you choose.

Choose wisely.

# Section 15: Gradient Descent

## The Simple Idea Behind Modern AI

Imagine you're hiking in thick fog. You can't see the valley ahead, but you can feel the slope beneath your feet.

Your strategy: take a step downhill. Feel the slope again. Take another step downhill. Repeat until the ground stops sloping—you've reached the valley.

This is exactly what **gradient descent** does.

It's an algorithm that finds the minimum of a function by:

1. **Starting** at a random point.
2. **Computing the gradient** (which way is downhill?).
3. **Taking a step** in the direction of the negative gradient.
4. **Repeating** until the loss stops decreasing.

The gradient tells us which direction to move.

Descending means following the negative gradient.

That's it. The entire algorithm is this simple.

It's the foundational algorithm for training nearly all modern machine learning models, from simple linear regression to the largest neural networks.

````python
import numpy as np
import matplotlib.pyplot as plt

# Simple function to minimize: f(x) = x²
f = lambda x: x**2
gradient = lambda x: 2*x

# Gradient descent parameters
x = 5.0  # Start at x = 5
learning_rate = 0.1
iterations = 20

# Track the path
path = [x]
losses = [f(x)]

# Perform gradient descent
for i in range(iterations):
    grad = gradient(x)
    x = x - learning_rate * grad  # Move in negative gradient direction
    path.append(x)
    losses.append(f(x))

# Visualization
x_range = np.linspace(-6, 6, 300)
y_range = f(x_range)

plt.figure(figsize=(8, 5))
plt.plot(x_range, y_range, linewidth=2, label='f(x) = x²')
plt.plot(path, losses, 'ro-', linewidth=2, markersize=6, label='Gradient Descent Path')
plt.scatter([path[0]], [losses[0]], color='green', s=150, zorder=5, label='Start')
plt.scatter([path[-1]], [losses[-1]], color='red', s=150, zorder=5, label='End')

plt.xlabel('x')
plt.ylabel('f(x)')
plt.title('Gradient Descent: Finding the Minimum')
plt.legend()
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

print("Starting point: x =", path[0])
print("Final point: x =", path[-1])
print("Starting loss:", losses[0])
print("Final loss:", losses[-1])
print(f"\nAfter {iterations} steps, we've reached very close to x = 0 (the minimum)")
````

## The Learning Rate: How Big Should Each Step Be?

The **learning rate** controls the size of each step downhill.

It's a hyperparameter you choose before training.

**Small learning rate (0.01):**
- ✅ Careful, precise updates
- ❌ Takes many iterations to converge
- ❌ Training is slow

**Large learning rate (1.0):**
- ✅ Fast progress initially
- ❌ Might overshoot the minimum
- ❌ Can diverge and make things worse

**Just right (0.1):**
- ✅ Converges efficiently
- ✅ Doesn't overshoot
- ✅ Balances speed and stability

Think of it like descending a mountain:

- Tiny steps: safe but slow.
- Giant steps: fast but you might miss the valley and tumble off the other side.
- Medium steps: the goldilocks zone.

In practice, finding the right learning rate is part of the art of machine learning.

Too small and training takes forever.

Too large and the algorithm diverges.

There are techniques (learning rate schedules) to automatically adjust this, but the core idea stays the same: balance speed with stability.

````python
def gradient_descent(f, gradient, x_start, learning_rate, iterations):
    """Run gradient descent and return the path."""
    x = x_start
    path = [x]
    losses = [f(x)]

    for _ in range(iterations):
        grad = gradient(x)
        x = x - learning_rate * grad
        path.append(x)
        losses.append(f(x))

    return np.array(path), np.array(losses)

# Test with different learning rates
f = lambda x: x**2
gradient = lambda x: 2*x

learning_rates = [0.01, 0.1, 0.5]
x_start = 5.0
iterations = 50

fig, axes = plt.subplots(1, 3, figsize=(14, 4))

x_range = np.linspace(-6, 6, 300)
y_range = f(x_range)

for ax, lr in zip(axes, learning_rates):
    path, losses = gradient_descent(f, gradient, x_start, lr, iterations)

    ax.plot(x_range, y_range, linewidth=2, color='blue')
    ax.plot(path, losses, 'ro-', linewidth=1.5, markersize=4, alpha=0.7)
    ax.scatter([path[0]], [losses[0]], color='green', s=100, zorder=5)
    ax.scatter([path[-1]], [losses[-1]], color='red', s=100, zorder=5)

    ax.set_xlabel('x')
    ax.set_ylabel('f(x)')
    ax.set_title(f'Learning Rate = {lr}')
    ax.grid(True, alpha=0.3)
    ax.set_ylim(-0.5, 26)

plt.tight_layout()
plt.show()

print("Learning Rate = 0.01:  Too small, converges slowly")
print("Learning Rate = 0.1:   Just right, efficient convergence")
print("Learning Rate = 0.5:   Still okay, but more oscillation")
````

## The Gradient Descent Algorithm

Here's the algorithm in plain language:

**Step 1:** Start at a random point (random x).

**Step 2:** Compute the gradient at that point.

**Step 3:** Take a step in the negative gradient direction: x_new = x_old - learning_rate × gradient

**Step 4:** Repeat Steps 2-3 until the loss stops improving.

That's it.

No magic.

No mysterious matrix inversions.

Just: compute slope, move downhill, repeat.

The beauty of gradient descent is its simplicity.

It works for any function where we can compute the gradient.

It works for functions with one variable or a million variables.

It works for simple functions or complex neural networks.

The only things that change are:

- The function (from simple parabola to complex loss surface).
- The gradient (computed by backpropagation in neural networks).
- The number of variables (from 1 to millions).

But the core algorithm is always the same.

````python
def numerical_gradient(f, x, h=1e-5):
    """Estimate gradient using finite differences."""
    return (f(x + h) - f(x)) / h

# Implement gradient descent from scratch
def gradient_descent_from_scratch(f, x_start, learning_rate, iterations):
    """Gradient descent using numerical gradients."""
    x = x_start

    print(f"{'Iteration':<12}{'x':<15}{'Loss':<15}")
    print("-" * 42)

    for i in range(iterations):
        loss = f(x)
        grad = numerical_gradient(f, x)

        if i % 5 == 0 or i == iterations - 1:
            print(f"{i:<12}{x:<15.6f}{loss:<15.6f}")

        x = x - learning_rate * grad

    return x, f(x)

# Test function
f = lambda x: (x - 3)**2 + 2

print("Minimizing f(x) = (x - 3)² + 2")
print("The minimum is at x = 3 with loss = 2\n")

x_final, loss_final = gradient_descent_from_scratch(
    f,
    x_start=0,
    learning_rate=0.1,
    iterations=50
)

print(f"\nFinal: x = {x_final:.6f}, loss = {loss_final:.6f}")
````

# Section 16: Computational Graphs

## How Frameworks Think About Computation

When you run code in PyTorch or TensorFlow, the framework doesn't immediately compute the result.

Instead, it builds a **computational graph**—a visual representation of how data flows through operations.

A computational graph is like a blueprint:

- **Nodes** represent operations (addition, multiplication, exponentiation, etc.).
- **Edges** represent data flowing between operations.
- **Arrows** show the direction of flow.

For example, consider this simple computation:

\[
y = (x + 2) \times 3
\]

You can break it into steps:

1. Add 2 to x.
2. Multiply the result by 3.

Each step is a node. The data flows through them in order.

This graph structure is powerful because:

- It shows exactly how data flows.
- It makes it easy to trace backward (which we need for backpropagation).
- It allows frameworks to optimize the computation.

In this section, we'll learn to think like a framework—to see every computation as a graph.

````python
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
from matplotlib.patches import FancyBboxPatch, FancyArrowPatch

# Create a simple computational graph for: y = (x + 2) * 3

fig, ax = plt.subplots(figsize=(10, 6))

# Define node positions
nodes = {
    'x': (1, 3),
    'const_2': (1, 1.5),
    'add': (3, 2.25),
    'const_3': (3, 0.75),
    'mul': (5, 1.5),
    'y': (7, 1.5)
}

# Draw nodes
node_colors = {
    'x': 'lightblue',
    'const_2': 'lightgreen',
    'add': 'lightyellow',
    'const_3': 'lightgreen',
    'mul': 'lightyellow',
    'y': 'lightcoral'
}

node_labels = {
    'x': 'x',
    'const_2': '2',
    'add': '+',
    'const_3': '3',
    'mul': '×',
    'y': 'y'
}

for node, (x, y) in nodes.items():
    box = FancyBboxPatch(
        (x - 0.3, y - 0.3), 0.6, 0.6,
        boxstyle="round,pad=0.05",
        edgecolor='black',
        facecolor=node_colors[node],
        linewidth=2
    )
    ax.add_patch(box)
    ax.text(x, y, node_labels[node], ha='center', va='center', fontsize=14, fontweight='bold')

# Draw edges (data flow)
edges = [
    ('x', 'add'),
    ('const_2', 'add'),
    ('add', 'mul'),
    ('const_3', 'mul'),
    ('mul', 'y')
]

for start, end in edges:
    x1, y1 = nodes[start]
    x2, y2 = nodes[end]

    arrow = FancyArrowPatch(
        (x1 + 0.3, y1), (x2 - 0.3, y2),
        arrowstyle='->', mutation_scale=20, linewidth=2, color='black'
    )
    ax.add_patch(arrow)

# Labels for intermediate results
ax.text(2, 2.7, 'x + 2', fontsize=10, style='italic', bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.5))
ax.text(4, 2, 'a', fontsize=10, style='italic', bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.5))
ax.text(6, 1.1, 'a × 3', fontsize=10, style='italic', bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.5))

ax.set_xlim(0, 8)
ax.set_ylim(0, 4)
ax.axis('off')
ax.set_aspect('equal')

plt.title('Computational Graph: y = (x + 2) × 3', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()

print("Computational Graph for: y = (x + 2) × 3")
print("\nNodes (operations): +, ×")
print("Edges (data flow): x → (+) → (×) → y")
print("\nThe graph shows how data flows through each operation.")
````

## Forward Pass and Backward Pass

A computational graph has two directions of flow:

### Forward Pass (Computing the Output)

Data flows from inputs to outputs.

For y = (x + 2) × 3 with x = 5:
Step 1: x = 5
Step 2: x + 2 = 7 (intermediate result)
Step 3: 7 × 3 = 21 (final result, y = 21)

This is what happens when you run the computation normally.

### Backward Pass (Computing Gradients)

After computing the output, we want to know:

> **How much did each input contribute to the final output?**

This requires flowing information **backward** through the graph.

We use the chain rule (from Section 12) to compute partial derivatives at each node.

For y = (x + 2) × 3:
dy/dy = 1 (starting point)
dy/d(x+2) = dy/dy × d(y)/d(x+2) = 1 × 3 = 3
dy/dx = dy/d(x+2) × d(x+2)/dx = 3 × 1 = 3

So if x changes by 1, y changes by 3. The gradient is 3.

Frameworks like PyTorch automate this backward flow using a process called **automatic differentiation** (which we'll cover in the next section).

But the core idea is simple:

- **Forward:** compute the output.
- **Backward:** compute how much each input mattered.

````python
# Trace through a computation step by step

x = 5

# Forward pass
print("=== FORWARD PASS ===")
print(f"Input: x = {x}")

step_1 = x + 2
print(f"Step 1 (add): x + 2 = {x} + 2 = {step_1}")

y = step_1 * 3
print(f"Step 2 (multiply): {step_1} × 3 = {y}")
print(f"Output: y = {y}\n")

# Backward pass
print("=== BACKWARD PASS ===")
print("We want: how much did x affect y?")
print()

# Start from the end
dy_dy = 1
print(f"dy/dy = {dy_dy} (starting point)")

# Work backward through each operation
dy_d_sum = dy_dy * 3  # derivative of multiplication
print(f"dy/d(x+2) = dy/dy × d(y)/d(x+2) = {dy_dy} × 3 = {dy_d_sum}")

dy_dx = dy_d_sum * 1  # derivative of addition
print(f"dy/dx = dy/d(x+2) × d(x+2)/dx = {dy_d_sum} × 1 = {dy_dx}")

print(f"\nConclusion: If x changes by 1, y changes by {dy_dx}")

# Verify numerically
x_new = x + 1
y_new = (x_new + 2) * 3
change_in_y = y_new - y
print(f"\nVerification: If x goes from {x} to {x_new}, y goes from {y} to {y_new}")
print(f"Change in y: {change_in_y} ✓")
````

# Section 17: Which Weight Should Change?

## The Central Question of Deep Learning

Imagine a neural network with thousands of weights.

After computing predictions, you realize the loss is high.

The natural question arises:

> **Which weight caused this error?**

> **Which weight should I adjust?**

> **How much should I adjust it?**

This section answers these questions.

It's the conceptual bridge between "I understand calculus" and "I understand how neural networks actually learn."

The answer involves everything we've learned:

- **Partial derivatives** tell us how much each weight affects the loss.
- **The chain rule** tells us how to compute those partial derivatives through layers.
- **Gradients** collect all those partial derivatives together.
- **Gradient descent** uses them to update weights.

But before we dive into the mechanics of backpropagation, let's build the intuition.

````python
import numpy as np
import matplotlib.pyplot as plt

# Simple example: a tiny neural network with 2 weights

# Forward pass
x = 2  # input
w1 = 0.5  # first weight
w2 = 0.3  # second weight
actual_output = 1.5  # ground truth

# Compute output
hidden = x * w1  # first layer: multiply by w1
predicted_output = hidden * w2  # second layer: multiply by w2

# Compute loss
loss = (predicted_output - actual_output)**2

print("=== FORWARD PASS ===")
print(f"Input: x = {x}")
print(f"Weight 1: w1 = {w1}")
print(f"Weight 2: w2 = {w2}")
print()
print(f"Hidden layer: x × w1 = {x} × {w1} = {hidden}")
print(f"Output: hidden × w2 = {hidden} × {w2} = {predicted_output}")
print(f"Actual output: {actual_output}")
print(f"Loss: (predicted - actual)² = ({predicted_output} - {actual_output})² = {loss}")
print()

print("=== THE QUESTION ===")
print("The loss is too high. Which weight should we change?")
print(f"Should we adjust w1 = {w1} or w2 = {w2}?")
print(f"Or both? By how much?")
````

## Tracing the Impact: How Does w1 Affect the Loss?

Let's trace how w1 impacts the final loss.

**w1 affects:**

→ Hidden value (hidden = x × w1)

→ Output (output = hidden × w2)

→ Loss (loss = (output - actual)²)

So there's a chain:
w1 → hidden → output → loss

To find how much w1 affects the loss, we use the chain rule:

\[
\frac{\partial \text{loss}}{\partial w_1} = \frac{\partial \text{loss}}{\partial \text{output}} \times \frac{\partial \text{output}}{\partial \text{hidden}} \times \frac{\partial \text{hidden}}{\partial w_1}
\]

Let's compute each piece:

**1. How much does the loss change with the output?**

\[
\frac{\partial \text{loss}}{\partial \text{output}} = 2 \times (\text{output} - \text{actual})
\]

**2. How much does the output change with the hidden value?**

\[
\frac{\partial \text{output}}{\partial \text{hidden}} = w_2
\]

**3. How much does the hidden value change with w1?**

\[
\frac{\partial \text{hidden}}{\partial w_1} = x
\]

Multiply them together and you get the total impact of w1 on the loss.

````python
# Compute partial derivatives step by step

print("=== BACKWARD PASS: COMPUTING GRADIENTS ===")
print()

# Step 1: How much does loss change with output?
dloss_doutput = 2 * (predicted_output - actual_output)
print(f"dloss/doutput = 2 × (output - actual)")
print(f"             = 2 × ({predicted_output} - {actual_output})")
print(f"             = {dloss_doutput}")
print()

# Step 2: How much does output change with hidden?
doutput_dhidden = w2
print(f"doutput/dhidden = w2 = {w2}")
print()

# Step 3: How much does hidden change with w1?
dhidden_dw1 = x
print(f"dhidden/dw1 = x = {x}")
print()

# Step 4: Chain them together
dloss_dw1 = dloss_doutput * doutput_dhidden * dhidden_dw1
print(f"dloss/dw1 = dloss/doutput × doutput/dhidden × dhidden/dw1")
print(f"          = {dloss_doutput} × {doutput_dhidden} × {dhidden_dw1}")
print(f"          = {dloss_dw1}")
print()

# Same for w2
doutput_dw2 = hidden
dloss_dw2 = dloss_doutput * doutput_dw2
print(f"dloss/dw2 = dloss/doutput × doutput/dw2")
print(f"          = {dloss_doutput} × {doutput_dw2}")
print(f"          = {dloss_dw2}")
````

## Which Weight Has More Impact?

Now we have the answer:

\[
\frac{\partial \text{loss}}{\partial w_1} = {dloss_dw1}
\]

\[
\frac{\partial \text{loss}}{\partial w_2} = {dloss_dw2}
\]

The larger the gradient, the more that weight affects the loss.

These gradients tell us:

- **Direction:** Is the gradient positive or negative? If positive, increasing the weight increases the loss (bad). If negative, increasing the weight decreases the loss (good).
- **Magnitude:** How much does the weight matter? A large magnitude means this weight has a big effect.

To reduce the loss, we move each weight in the negative gradient direction:

\[
w_1 \leftarrow w_1 - \text{learning\_rate} \times \frac{\partial \text{loss}}{\partial w_1}
\]

\[
w_2 \leftarrow w_2 - \text{learning\_rate} \times \frac{\partial \text{loss}}{\partial w_2}
\]

This is exactly what gradient descent does—it updates weights based on how much they contributed to the error.

````python
# Update the weights using gradient descent

learning_rate = 0.1

print("=== UPDATING WEIGHTS ===")
print()
print(f"Learning rate: {learning_rate}")
print()
print(f"Old w1: {w1}")
print(f"w1 ← w1 - learning_rate × dloss/dw1")
print(f"   ← {w1} - {learning_rate} × {dloss_dw1}")
w1_new = w1 - learning_rate * dloss_dw1
print(f"   ← {w1_new}")
print()

print(f"Old w2: {w2}")
print(f"w2 ← w2 - learning_rate × dloss/dw2")
print(f"   ← {w2} - {learning_rate} × {dloss_dw2}")
w2_new = w2 - learning_rate * dloss_dw2
print(f"   ← {w2_new}")
print()

# Compute new loss with updated weights
hidden_new = x * w1_new
predicted_new = hidden_new * w2_new
loss_new = (predicted_new - actual_output)**2

print(f"Old loss: {loss}")
print(f"New loss: {loss_new}")
print(f"Improvement: {loss - loss_new}")
````

## The Key Insight

Every weight has a partial derivative with respect to the loss.

That partial derivative tells us:

1. **How much does this weight matter?** (magnitude)
2. **In which direction should we adjust it?** (sign)

In a real neural network, there are millions of weights.

But the principle is exactly the same:

- Compute the loss.
- Compute how each weight affected the loss (using partial derivatives and the chain rule).
- Update each weight to reduce the loss.

This is the foundation of training all neural networks.

The process of computing these partial derivatives through the network is called **backpropagation**.

It's just the chain rule applied systematically to every weight.

# Section 18: Backpropagation

## What Is Backpropagation?

Backpropagation is the process of computing how much each weight in a neural network contributed to the final error.

It does this by flowing gradients backward through the network, applying the chain rule at each step.

The name says it all: "back" + "propagation" = sending information backward through the layers.

Here's the core idea:

1. **Forward pass:** Data flows through the network, producing a prediction and a loss.
2. **Backward pass:** Gradients flow backward, computing how each weight affected the loss.
3. **Update weights:** Each weight is adjusted based on its gradient.

That's backpropagation.

It's not a new algorithm. It's the chain rule applied systematically to a neural network.

Everything in this workbook has been building toward understanding this one process:

- Derivatives measure how things change.
- Partial derivatives measure change in one direction.
- The chain rule connects changes across multiple steps.
- Gradients collect all partial derivatives.
- Backpropagation applies all of this to train neural networks.

## The Backpropagation Algorithm

Backpropagation is systematic application of the chain rule.

Here's the algorithm in plain terms:

**Step 1: Forward Pass**

Compute predictions by passing data through the network layer by layer.

Compute the loss.

**Step 2: Start at the End**

Compute the gradient of the loss with respect to the output.
∂Loss/∂output = 2 × (output - actual)

**Step 3: Flow Backward**

For each layer going backward, use the chain rule:
∂Loss/∂(previous layer) = ∂Loss/∂(current output) × ∂(current output)/∂(previous layer)

**Step 4: Compute Weight Gradients**

For each weight in each layer:
∂Loss/∂weight = ∂Loss/∂(output of this layer) × ∂(output)/∂weight

**Step 5: Update Weights**
weight ← weight - learning_rate × ∂Loss/∂weight

The beauty of backpropagation is that it's completely systematic.

You don't need to think about the specific network structure.

The chain rule handles everything automatically.

It's why backpropagation works for networks of any depth or complexity.

````python
# Trace backpropagation through a 3-layer network

print("=== BACKPROPAGATION EXAMPLE ===\n")

# Setup: tiny network
x = 1.0
w1 = 2.0
w2 = 3.0
actual_y = 10.0

print("Network: input → (×w₁) → hidden → (×w₂) → output")
print(f"Input: x = {x}")
print(f"Weights: w₁ = {w1}, w₂ = {w2}")
print(f"Actual output: y = {actual_y}\n")

# FORWARD PASS
print("=== FORWARD PASS ===")
hidden = x * w1
output = hidden * w2
loss = (output - actual_y)**2

print(f"hidden = x × w₁ = {x} × {w1} = {hidden}")
print(f"output = hidden × w₂ = {hidden} × {w2} = {output}")
print(f"loss = (output - actual_y)² = ({output} - {actual_y})² = {loss}\n")

# BACKWARD PASS
print("=== BACKWARD PASS ===")

# Start from the end
dloss_doutput = 2 * (output - actual_y)
print(f"Step 1: ∂loss/∂output = 2 × (output - actual_y)")
print(f"                      = 2 × ({output} - {actual_y})")
print(f"                      = {dloss_doutput}\n")

# Flow back to w2
doutput_dw2 = hidden
dloss_dw2 = dloss_doutput * doutput_dw2
print(f"Step 2: ∂loss/∂w₂ = ∂loss/∂output × ∂output/∂w₂")
print(f"                  = {dloss_doutput} × {doutput_dw2}")
print(f"                  = {dloss_dw2}\n")

# Flow back to hidden
doutput_dhidden = w2
dloss_dhidden = dloss_doutput * doutput_dhidden
print(f"Step 3: ∂loss/∂hidden = ∂loss/∂output × ∂output/∂hidden")
print(f"                      = {dloss_doutput} × {doutput_dhidden}")
print(f"                      = {dloss_dhidden}\n")

# Flow back to w1
dhidden_dw1 = x
dloss_dw1 = dloss_dhidden * dhidden_dw1
print(f"Step 4: ∂loss/∂w₁ = ∂loss/∂hidden × ∂hidden/∂w₁")
print(f"                  = {dloss_dhidden} × {dhidden_dw1}")
print(f"                  = {dloss_dw1}\n")

# UPDATE WEIGHTS
print("=== WEIGHT UPDATE ===")
learning_rate = 0.01

w1_new = w1 - learning_rate * dloss_dw1
w2_new = w2 - learning_rate * dloss_dw2

print(f"w₁ ← w₁ - learning_rate × ∂loss/∂w₁")
print(f"   ← {w1} - {learning_rate} × {dloss_dw1}")
print(f"   ← {w1_new}\n")

print(f"w₂ ← w₂ - learning_rate × ∂loss/∂w₂")
print(f"   ← {w2} - {learning_rate} × {dloss_dw2}")
print(f"   ← {w2_new}\n")

# New loss
hidden_new = x * w1_new
output_new = hidden_new * w2_new
loss_new = (output_new - actual_y)**2

print(f"Old loss: {loss}")
print(f"New loss: {loss_new}")
print(f"Improvement: {loss - loss_new}")
````

## Why It Works: The Chain Rule at Scale

Backpropagation is powerful because it scales.

A simple neural network might have:

- 1,000 weights in the first layer
- 1,000 weights in the second layer
- 1,000 weights in the third layer

That's **3,000,000 weights**.

Backpropagation computes gradients for all of them simultaneously using the chain rule.

The computation flows backward:
∂Loss/∂w_last_layer
↓ (chain rule)
∂Loss/∂w_middle_layer
↓ (chain rule)
∂Loss/∂w_first_layer

Each step is just applying the chain rule.

Because the chain rule works for any number of layers, backpropagation works for deep networks.

This is why deep learning became possible:

1. We had the math (backpropagation, discovered in the 1980s).
2. We had the data (big datasets).
3. We had the compute (GPUs, discovered to be fast at neural network training).

Without backpropagation, training deep networks would be impossibly slow.

# Section 19: Automatic Differentiation

## From Manual Math to Automatic Computation

In the previous sections, we computed gradients by hand using the chain rule.

We traced through the network layer by layer, computing partial derivatives at each step.

It works, but it's tedious and error-prone.

A better question: **Can we make computers do this automatically?**

The answer is yes. It's called **automatic differentiation** (AD).

Automatic differentiation is a technique that computes exact derivatives of arbitrary functions by systematically applying the chain rule.

It's not numerical approximation (like finite differences).

It's not symbolic manipulation (like a computer algebra system).

It's a clever algorithm that tracks how data flows through operations and automatically computes gradients.

This is how modern frameworks like PyTorch, TensorFlow, and JAX work.

When you call `.backward()`, you're invoking automatic differentiation.

In this section, we'll understand how it works.

````python
import torch

# Simple example: automatic differentiation in PyTorch

x = torch.tensor(2.0, requires_grad=True)
w = torch.tensor(3.0, requires_grad=True)

print("=== AUTOMATIC DIFFERENTIATION ===\n")
print(f"x = {x.item()}")
print(f"w = {w.item()}")
print()

# Forward pass
y = x * w
print(f"y = x × w = {y.item()}\n")

# Backward pass (automatic!)
y.backward()

print("After y.backward():")
print(f"∂y/∂x = {x.grad.item()}")
print(f"∂y/∂w = {w.grad.item()}")
print()

print("PyTorch computed these gradients automatically!")
print("We didn't write any chain rule code.")
print("The framework did it for us.")
````

## How Automatic Differentiation Works

Automatic differentiation works by tracking how data flows through operations.

For each operation, the framework knows the derivative.

For example:

- Derivative of (a + b): ∂/∂a = 1, ∂/∂b = 1
- Derivative of (a × b): ∂/∂a = b, ∂/∂b = a
- Derivative of sin(x): ∂/∂x = cos(x)
- Derivative of exp(x): ∂/∂x = exp(x)

During the forward pass, the framework doesn't just compute results.

It builds a **computational graph** recording which operations were used.

During the backward pass, it traverses this graph in reverse, applying the chain rule at each step using the known derivatives.

The key insight: every complex function is just a combination of simple operations.

If you know the derivatives of the simple operations, you can compute the derivative of the complex function using the chain rule.

That's what automatic differentiation does.

````python
# More complex example: automatic differentiation through multiple operations

x = torch.tensor(2.0, requires_grad=True)

print("=== COMPLEX FUNCTION ===\n")

# Compute y = (x² + 3x) × sin(x)
y = (x**2 + 3*x) * torch.sin(x)

print(f"x = {x.item()}")
print(f"y = (x² + 3x) × sin(x) = {y.item()}\n")

# Backward pass
y.backward()

print(f"∂y/∂x = {x.grad.item()}\n")

print("This gradient was computed automatically!")
print("PyTorch applied the chain rule through:")
print("  - Squaring")
print("  - Multiplication by 3")
print("  - Addition")
print("  - Multiplication by sin(x)")
print("  - The sine operation")
print()
print("All without us writing a single chain rule!")

# Verify numerically
h = 1e-5
x_plus = 2.0 + h
y_plus = (x_plus**2 + 3*x_plus) * np.sin(x_plus)
numerical_grad = (y_plus - y.item()) / h

print(f"\nNumerical gradient (finite differences): {numerical_grad:.6f}")
print(f"Automatic differentiation gradient:     {x.grad.item():.6f}")
print("They match!")

import numpy as np
````

## Two Modes of Automatic Differentiation

Automatic differentiation can work in two directions.

### Reverse Mode (Backpropagation)

Start at the output and flow gradients backward.

This is what neural networks use.

**Advantages:**

- One backward pass computes gradients for all inputs.
- Efficient when you have one output and many inputs (neural networks).

**Disadvantages:**

- Requires storing the computational graph during forward pass.
- Memory overhead.

### Forward Mode

Start at the input and flow derivatives forward.

**Advantages:**

- No need to store the full computational graph.
- Efficient when you have many outputs and few inputs.

**Disadvantages:**

- Need one forward pass per input (slow if many inputs).

Neural networks use **reverse mode** automatic differentiation because:

- Loss is one scalar output.
- You have many weights (inputs).
- One backward pass computes gradients for all weights.

This is why backpropagation and reverse-mode AD are closely related—they're describing the same process from different angles.

## Building Gradient Descent from Scratch

Here's the algorithm we've been studying, implemented without any deep learning framework:

```python
def gradient_descent(f, gradient_f, x_start, learning_rate, iterations):
    """
    Minimize f(x) using gradient descent.
    
    f: the function to minimize
    gradient_f: function that computes ∂f/∂x
    x_start: starting point
    learning_rate: step size
    iterations: number of steps
    """
    x = x_start
    history = [x]
    
    for i in range(iterations):
        grad = gradient_f(x)
        x = x - learning_rate * grad
        history.append(x)
    
    return x, history
```

That's it.

This simple algorithm—repeat by repeat, step by step—powers modern AI.

The only difference between this and training a billion-parameter neural network is:

- The function is more complex (neural network loss instead of f(x) = x²)
- The gradient is computed via backpropagation (instead of an analytic formula)
- There are millions of variables instead of one

But the core algorithm is identical.

````python
import numpy as np
import matplotlib.pyplot as plt

def gradient_descent(f, gradient_f, x_start, learning_rate, iterations):
    """Minimize f(x) using gradient descent."""
    x = x_start
    history = [x]
    losses = [f(x)]

    for i in range(iterations):
        grad = gradient_f(x)
        x = x - learning_rate * grad
        history.append(x)
        losses.append(f(x))

    return x, history, losses

# Example 1: Minimize f(x) = x²
print("=== EXAMPLE 1: f(x) = x² ===\n")

f1 = lambda x: x**2
grad_f1 = lambda x: 2*x

x_final, history, losses = gradient_descent(
    f1, grad_f1,
    x_start=5.0,
    learning_rate=0.1,
    iterations=50
)

print(f"Starting point: x = {history[0]}")
print(f"Final point: x = {x_final:.6f}")
print(f"Starting loss: {losses[0]:.6f}")
print(f"Final loss: {losses[-1]:.6f}\n")

# Example 2: Minimize f(x) = (x - 3)² + 2
print("=== EXAMPLE 2: f(x) = (x - 3)² + 2 ===\n")

f2 = lambda x: (x - 3)**2 + 2
grad_f2 = lambda x: 2*(x - 3)

x_final2, history2, losses2 = gradient_descent(
    f2, grad_f2,
    x_start=0.0,
    learning_rate=0.1,
    iterations=50
)

print(f"Starting point: x = {history2[0]}")
print(f"Final point: x = {x_final2:.6f} (should be near 3)")
print(f"Starting loss: {losses2[0]:.6f}")
print(f"Final loss: {losses2[-1]:.6f}\n")

# Visualization
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

# Plot 1
x_range = np.linspace(-6, 6, 300)
axes[0].plot(x_range, f1(x_range), linewidth=2, label='f(x) = x²')
axes[0].plot(history, losses, 'ro-', linewidth=2, markersize=5, label='Gradient descent path')
axes[0].scatter([history[0]], [losses[0]], color='green', s=100, zorder=5, label='Start')
axes[0].scatter([x_final], [losses[-1]], color='red', s=100, zorder=5, label='End')
axes[0].set_xlabel('x')
axes[0].set_ylabel('f(x)')
axes[0].set_title('Minimizing f(x) = x²')
axes[0].legend()
axes[0].grid(True, alpha=0.3)

# Plot 2
x_range2 = np.linspace(-2, 8, 300)
axes[1].plot(x_range2, f2(x_range2), linewidth=2, label='f(x) = (x-3)² + 2')
axes[1].plot(history2, losses2, 'ro-', linewidth=2, markersize=5, label='Gradient descent path')
axes[1].scatter([history2[0]], [losses2[0]], color='green', s=100, zorder=5, label='Start')
axes[1].scatter([x_final2], [losses2[-1]], color='red', s=100, zorder=5, label='End')
axes[1].set_xlabel('x')
axes[1].set_ylabel('f(x)')
axes[1].set_title('Minimizing f(x) = (x-3)² + 2')
axes[1].legend()
axes[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

print("Both examples converge to their minima!")
print("The same algorithm works for any differentiable function.")
````

This is why understanding gradient descent matters.

It's not just an algorithm. It's the foundation of all modern machine learning.

````python
import torch
import torch.nn as nn

print("=== PYTORCH GRADIENT DESCENT ===\n")

# Simple linear model: y = w*x + b
model = nn.Linear(1, 1)
optimizer = torch.optim.SGD(model.parameters(), lr=0.01)
loss_fn = nn.MSELoss()

# Training data
x_data = torch.tensor([[1.0], [2.0], [3.0], [4.0]])
y_data = torch.tensor([[2.0], [4.0], [6.0], [8.0]])

print("Training a simple linear model: y = w*x + b\n")
print(f"{'Epoch':<8}{'Loss':<15}")
print("-" * 23)

# Training loop (5 iterations of gradient descent)
for epoch in range(5):
    # Forward pass
    y_pred = model(x_data)
    loss = loss_fn(y_pred, y_data)

    # Backward pass (compute gradients)
    optimizer.zero_grad()  # Reset gradients
    loss.backward()         # Backpropagation

    # Update weights (gradient descent step)
    optimizer.step()

    print(f"{epoch:<8}{loss.item():<15.6f}")

print("\nPyTorch did all the calculus for us!")
print("Behind the scenes: backpropagation → gradient descent → parameter update")

# Show final weights
w, b = model.parameters()
print(f"\nLearned weights:")
print(f"w ≈ {w.item():.4f} (should be close to 2)")
print(f"b ≈ {b.item():.4f} (should be close to 0)")
````

import numpy as np

x = np.array([1, 2, 3, 4])

w1 = np.random.randn()

for epoch in range(10):
    # Forward pass
    hidden = x * w1
